```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/vincentarelbundock/Rdatasets/master/csv/AER/TeachingRatings.csv"
df = pd.read_csv(url)
df.head()
```

```python
minority_df = df[df['minority'] == 'yes']
pct_tenured = (minority_df['tenure'] == 'yes').mean() * 100
print("Tenured % among minority professors:", pct_tenured)

pd.crosstab(df['minority'], df['tenure'], normalize='index') * 100
```

84.38% of minority professors are tenured vs 76.94% of non-minority professors a small difference, not a strong one.therefore, the tenure rate is higher aqmong visible minority professors by about 7.44% suggesting that tenure status differs by minority status

```python
df.groupby('tenure')['age'].agg(['mean', 'std'])
```

Tenured instructors have a higher mean age (47.85years) than non-tenured instructors (50.18 years).

```python
df['age'].hist(bins=15, edgecolor='black')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
```

histogram works better for the age variable because age is a continuous numerical values. a histogram shows how the ages are distributed across different intervals.most professors are approx 45 to 55 yrs old.

```python
gender_counts=df['gender'].value_counts()
print(gender_counts)
gender_counts.plot(kind='bar', color=['blue','orange'], edgecolor='black')
plt.title('Gender Count')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.show()
```

```python
plt.barh(gender_counts.index, gender_counts.values, color=['blue', 'orange'])
plt.title('Gender Count')
plt.xlabel('Count')
plt.ylabel('Gender')
plt.show()
```

plt.bar()-vertical bars (categories on x-axis)
plt.barh()-horizontal bars (categories on y-axis). it is better when labels are long

```python
tenured=df[df['tenure']=='yes']
median_eval=tenured["eval"].median()
print(median_eval)
```

a median of 4.0 suggests tenured professors are generally rated fairly well by students right around good and not average or poor
